```python
df = df.drop(columns=["race/ethnicity", "parental level of education", "lunch", "test preparation course"])
```

```python
df.shape
```

```python
mean_math = df["math score"].mean()
df = df[df["math score"] <= mean_math]
```

```python
df.shape
```

```python
df.head()

df[df["math score"]] > 16
```

```python
import matplotlib.pyplot as mlt
print("done")
```

```python
mlt.figure(figsize=(10,10))
mlt.bar(df.index, df["math score"], color="orange", edgecolor="black")
mlt.axhline(df["math score"].mean(), color="red", linestyle="dashed", linewidth=2, label=f"Avg = {df["math score"].mean():.2f}")
mlt.title("Students Below Average Math Score")
mlt.xlabel("Student ID")
mlt.ylabel("Math Score")
mlt.legend()
mlt.show()
```
