#7.Calculate minimum, maximum, range, variance, standard deviation, quartiles, interquartile range (IQR), and coefficient of variation (where meaningful).

```python
df.describe()
```

```python
Q1_profit = df['Sales'].quantile(0.25)
Q3_profit = df['Sales'].quantile(0.75)
IQR_profit = Q3_profit - Q1_profit
print(IQR_profit)
```

```python
Q1_profit = df['Profit'].quantile(0.25)
Q3_profit = df['Profit'].quantile(0.75)
IQR_profit = Q3_profit - Q1_profit
print(IQR_profit)
```

```python
Q1_profit = df['Margin'].quantile(0.25)
Q3_profit = df['Margin'].quantile(0.75)
IQR_profit = Q3_profit - Q1_profit
print(IQR_profit)
```

```python
sales_by_product_type = df.groupby('Product_type')['Sales'].mean()   
```

```python
print(sales_by_product_type)
```

```python
sales_mode_by_product_type = df.groupby('Product_type')['Sales'].agg(lambda x: x.mode().iloc[0])
```

```python
print(sales_mode_by_product_type)
```

```python
sales_median_by_product_type = df.groupby('Product_type')['Sales'].median()
print(sales_median_by_product_type)
```

8.Identify the distribution of at least three numerical columns using numerical summaries and graphical analysis. Comment on skewness and possible normality/uniformity.

Plot target sales by current sales 

```python
plt.plot(df['Date'], df['Profit'], label='Profit')
plt.plot(df['Date'], df['Sales'], label='Sales')
plt.xlabel('Date')
plt.ylabel('Sales')
plt.title('Profit vs Sales Over Time')
plt.legend()
plt.show()   
```

```python


plt.scatter(df['Profit'], df['Profit'])
plt.plot([0, df[['Profit','Sales']].max().max()],
         [0, df[['Profit','Sales']].max().max()],
         'r--', label='1:1 line')  # ideal line
plt.xlabel('Profit')
plt.ylabel('Sales')
plt.title('Profit vs Sales')
plt.legend()
plt.show()   
```
